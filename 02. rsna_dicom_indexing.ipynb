Xây file index (manifest) từ DICOM + metadata, dùng cho các bước EDA/training sau. Tiền đề: đã chạy `01. rsna_eda_metadata.ipynb` và xác nhận schema thật.

```python
import sys, os
sys.path.append('..')
import pandas as pd
from rsna_data import get_dataset_root, list_dicom_files
from rsna_lib import RANDOM_SEED

ROOT = get_dataset_root()
```

## Quét DICOM theo batch (tránh load toàn bộ pixel data vào RAM)

```python
dicom_paths = list_dicom_files(ROOT)
print(f'Tổng số file DICOM: {len(dicom_paths)}')
```

## Đọc header DICOM (không đọc pixel array) để lấy metadata nhẹ

```python
import pydicom

records = []
for p in dicom_paths[:5]:  # TODO: bỏ giới hạn [:5] sau khi kiểm thử schema đúng
    ds = pydicom.dcmread(p, stop_before_pixels=True)
    records.append({
        'path': p,
        'StudyInstanceUID': getattr(ds, 'StudyInstanceUID', None),
        'SeriesInstanceUID': getattr(ds, 'SeriesInstanceUID', None),
    })
index_df = pd.DataFrame(records)
index_df.head()
```
